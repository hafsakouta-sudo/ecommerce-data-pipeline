```python
geolocation = pd.read_csv(
    "../data/raw/olist_geolocation_dataset.csv"
)

geolocation.head()
```

```python
geolocation.shape
```

```python
geolocation.isnull().sum()
```

```python
geolocation.duplicated().sum()
```

```python
geolocation.dtypes
```

```python
geolocation = geolocation.drop_duplicates()
```

```python
geolocation.duplicated().sum()
```

```python
geolocation["geolocation_zip_code_prefix"].nunique()
```

```python
len(geolocation)
```

```python
geolocation["geolocation_zip_code_prefix"].value_counts().head(10)
```

```python
geolocation_cleaned = (
    geolocation
    .groupby("geolocation_zip_code_prefix")
    .agg({
        "geolocation_lat": "mean",
        "geolocation_lng": "mean"
    })
    .reset_index()
)
```

```python
geolocation_cleaned.shape
```

```python
geolocation.groupby(
    "geolocation_zip_code_prefix"
)["geolocation_city"].nunique().sort_values(ascending=False).head(10)
```

```python
geolocation[
    geolocation["geolocation_zip_code_prefix"] == 17970
]["geolocation_city"].value_counts()
```

```python
city_by_zip = (
    geolocation
    .groupby("geolocation_zip_code_prefix")["geolocation_city"]
    .agg(lambda x: x.value_counts().index[0])
    .reset_index()
)
```

```python
city_by_zip.head()
```

```python
state_by_zip = (
    geolocation
    .groupby("geolocation_zip_code_prefix")["geolocation_state"]
    .agg(lambda x: x.value_counts().index[0])
    .reset_index()
)
```

```python
state_by_zip.head()
```

```python
geolocation_cleaned = geolocation_cleaned.merge(
    city_by_zip,
    on="geolocation_zip_code_prefix",
    how="left"
)

geolocation_cleaned = geolocation_cleaned.merge(
    state_by_zip,
    on="geolocation_zip_code_prefix",
    how="left"
)
```

```python
geolocation_cleaned.head()
```

```python
geolocation_cleaned.shape
```

```python
geolocation_cleaned["geolocation_zip_code_prefix"].duplicated().sum()
```

```python
geolocation_cleaned.to_csv(
    "../data/processed/geolocation_cleaned.csv",
    index=False
)
```
